## Mathematical operations

1. Scalar Operation

```python
x = torch.rand(2,2)
x
```

```python
print(x)

# addition
print(x + 2)

# substraction
print(x - 2)

# multiplication
print(x * 2)

# division
print(x / 2)

# int division
print((x*100) // 2)

# mod
print(x % 2)

# power
print((x*100) ** 2)



```

2. elementwise operation

```python
a = torch.rand(2,3)
b = torch.rand(2,3)

print(a)
print(b)
```

```python
# addition
print(a + b)

# substraction
print(a - b)

# multiplication
print(a * b)

# division
print(a / b)

# int division
print((a*100) // b)

# mod
print(a % b)

# power
print((a*100) ** b)

```

```python
c = torch.tensor([1,-2,3,-4])
c
```

```python
# negative
torch.neg(c)
```

```python
d = torch.tensor([1.9, 2.3, 3.7, 4.4])
```

```python
# ceil
torch.ceil(d)
```

```python
# floor
torch.floor(d)
```

```python
# clamp
torch.clamp(d, min=2, max=3)
```

3. Reduction operation

```python
e = torch.randint(size=(2,3),low=0, high=10, dtype=torch.float32)
e
```

```python
# sum 
print(torch.sum(e))

# sum along columns
print(torch.sum(e, dim=0))

# sum along columns row
print(torch.sum(e, dim=1))
```

```python
# mean
print(torch.mean(e))

# mean
print(torch.mean(e, dim=0))

# mean along row
print(torch.mean(e, dim=1))
```

```python
# median
print(torch.median(e))
```

```python
print(torch.max(e))
print(torch.min(e))
```

```python
# product
torch.prod(e)
```

```python
# standard deviation
torch.std(e)
```

```python
# variance
torch.var(e)
```

```python
# argmax
torch.argmax(e)
```

```python
# argmin
torch.argmin(e)
```
